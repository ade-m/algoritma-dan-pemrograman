# Bagian A — Bubble Sort

## 2. Intuisi Bubble Sort

Bubble Sort membandingkan **dua elemen yang bersebelahan**.

Jika urutannya salah, keduanya ditukar.

Contoh:

```text
[5, 3, 8, 4]

Bandingkan 5 dan 3
5 > 3 → tukar

[3, 5, 8, 4]

Bandingkan 5 dan 8
tidak perlu tukar

[3, 5, 8, 4]

Bandingkan 8 dan 4
8 > 4 → tukar

[3, 5, 4, 8]
```

Setelah satu putaran, elemen terbesar akan "mengambang" ke bagian kanan.

````python
def bubble_sort_trace(data):
    arr = data.copy()
    n = len(arr)

    print("Data awal:", arr)
    print()

    for i in range(n - 1):
        print(f"PASS {i + 1}")

        for j in range(n - 1 - i):
            print(f"  Bandingkan {arr[j]} dan {arr[j + 1]}", end="")

            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                print(" -> tukar")
            else:
                print(" -> tetap")

            print(" ", arr)

        print()

    return arr


hasil = bubble_sort_trace([5, 3, 8, 4])
print("Hasil akhir:", hasil)
````

### Yang perlu diamati

Pada setiap **pass**:

- bagian kanan semakin pasti sudah terurut;
- jumlah perbandingan dapat dikurangi;
- elemen terbesar dari bagian yang belum terurut bergerak ke kanan.

## 3. Implementasi Bubble Sort

````python
def bubble_sort(data):
    arr = data.copy()
    n = len(arr)

    for i in range(n - 1):
        for j in range(n - 1 - i):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

    return arr


data = [78, 65, 90, 55, 88, 72]
print("Sebelum:", data)
print("Sesudah:", bubble_sort(data))
````

## 4. Bubble Sort dengan Early Stop

Bagaimana jika data ternyata sudah terurut?

Bubble Sort biasa tetap menjalankan banyak perbandingan.

Kita dapat menambahkan flag `swapped`.

Jika dalam satu pass **tidak ada pertukaran**, berarti data sudah terurut dan proses bisa dihentikan.

````python
def bubble_sort_optimized(data):
    arr = data.copy()
    n = len(arr)

    for i in range(n - 1):
        swapped = False

        for j in range(n - 1 - i):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True

        if not swapped:
            break

    return arr


print(bubble_sort_optimized([1, 2, 3, 4, 5]))
print(bubble_sort_optimized([5, 4, 3, 2, 1]))
````

## 5. Menghitung Operasi Bubble Sort

Mari hitung jumlah **comparison** dan **swap**.

````python
def bubble_sort_count(data):
    arr = data.copy()
    n = len(arr)

    comparisons = 0
    swaps = 0

    for i in range(n - 1):
        for j in range(n - 1 - i):
            comparisons += 1

            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swaps += 1

    return arr, comparisons, swaps


samples = {
    "Sudah terurut": [1, 2, 3, 4, 5],
    "Acak": [4, 2, 5, 1, 3],
    "Terbalik": [5, 4, 3, 2, 1]
}

for label, data in samples.items():
    hasil, comparisons, swaps = bubble_sort_count(data)
    print(label)
    print("  hasil       :", hasil)
    print("  comparisons :", comparisons)
    print("  swaps       :", swaps)
    print()
````

# Bagian B — Selection Sort

## 6. Intuisi Selection Sort

Selection Sort memilih **elemen terkecil** dari bagian yang belum terurut, lalu menempatkannya ke posisi yang seharusnya.

Contoh:

```text
[5, 3, 8, 4]

Cari nilai terkecil → 3
Tukar dengan posisi pertama

[3, 5, 8, 4]

Cari nilai terkecil dari sisa data → 4
Tukar dengan posisi kedua

[3, 4, 8, 5]

Cari nilai terkecil dari sisa data → 5

[3, 4, 5, 8]
```

````python
def selection_sort_trace(data):
    arr = data.copy()
    n = len(arr)

    print("Data awal:", arr)
    print()

    for i in range(n - 1):
        min_index = i

        print(f"PASS {i + 1}")
        print(f"  Kandidat awal minimum: {arr[min_index]}")

        for j in range(i + 1, n):
            print(f"  Bandingkan minimum {arr[min_index]} dengan {arr[j]}")

            if arr[j] < arr[min_index]:
                min_index = j
                print(f"    -> minimum baru: {arr[min_index]}")

        if min_index != i:
            arr[i], arr[min_index] = arr[min_index], arr[i]
            print(f"  Tukar posisi {i} dengan {min_index}")

        print(" ", arr)
        print()

    return arr


hasil = selection_sort_trace([5, 3, 8, 4])
print("Hasil akhir:", hasil)
````

## 7. Implementasi Selection Sort

````python
def selection_sort(data):
    arr = data.copy()
    n = len(arr)

    for i in range(n - 1):
        min_index = i

        for j in range(i + 1, n):
            if arr[j] < arr[min_index]:
                min_index = j

        arr[i], arr[min_index] = arr[min_index], arr[i]

    return arr


data = [78, 65, 90, 55, 88, 72]
print("Sebelum:", data)
print("Sesudah:", selection_sort(data))
````

## 8. Menghitung Operasi Selection Sort

Selection Sort tetap mencari minimum pada sisa data di setiap pass.

Mari hitung jumlah perbandingan dan pertukarannya.

````python
def selection_sort_count(data):
    arr = data.copy()
    n = len(arr)

    comparisons = 0
    swaps = 0

    for i in range(n - 1):
        min_index = i

        for j in range(i + 1, n):
            comparisons += 1

            if arr[j] < arr[min_index]:
                min_index = j

        if min_index != i:
            arr[i], arr[min_index] = arr[min_index], arr[i]
            swaps += 1

    return arr, comparisons, swaps


for label, data in samples.items():
    hasil, comparisons, swaps = selection_sort_count(data)
    print(label)
    print("  hasil       :", hasil)
    print("  comparisons :", comparisons)
    print("  swaps       :", swaps)
    print()
````

# Bagian C — Membandingkan Keduanya

## 9. Bubble Sort vs Selection Sort

| Aspek | Bubble Sort | Selection Sort |
|---|---|---|
| Ide utama | Tukar elemen bersebelahan | Pilih minimum |
| Perbandingan | Banyak | Banyak |
| Pertukaran | Bisa sangat banyak | Relatif lebih sedikit |
| Best case | Bisa O(n) jika dioptimasi | O(n²) |
| Average case | O(n²) | O(n²) |
| Worst case | O(n²) | O(n²) |
| Cocok untuk belajar | Ya | Ya |

### Dari mana O(n²) berasal?

Jumlah perbandingan kurang lebih:

```text
(n - 1) + (n - 2) + ... + 1
```

Rumusnya:

\[
\frac{n(n-1)}{2}
\]

Jika bagian dominannya diambil, pertumbuhannya adalah:

\[
O(n^2)
\]

````python
def estimasi_comparisons(n):
    return n * (n - 1) // 2

for n in [10, 100, 1_000, 10_000]:
    print(f"n={n:>6,} -> sekitar {estimasi_comparisons(n):>12,} comparison")
````

## 10. Eksperimen Waktu Eksekusi

Kita bandingkan waktu Bubble Sort dan Selection Sort pada data acak.

> Catatan: hasil waktu dapat berbeda antar komputer.

````python
import random
import time

def benchmark(sort_function, data):
    start = time.perf_counter()
    sort_function(data)
    end = time.perf_counter()
    return end - start


sizes = [100, 500, 1000, 2000]

for n in sizes:
    data = [random.randint(1, 1_000_000) for _ in range(n)]

    bubble_time = benchmark(bubble_sort, data)
    selection_time = benchmark(selection_sort, data)

    print(
        f"n={n:>4} | "
        f"Bubble={bubble_time:.6f}s | "
        f"Selection={selection_time:.6f}s"
    )
````

````python
import matplotlib.pyplot as plt
import random
import time

sizes = [100, 300, 500, 800, 1000]
bubble_times = []
selection_times = []

for n in sizes:
    data = [random.randint(1, 1_000_000) for _ in range(n)]

    bubble_times.append(benchmark(bubble_sort, data))
    selection_times.append(benchmark(selection_sort, data))

plt.figure(figsize=(8, 5))
plt.plot(sizes, bubble_times, marker="o", label="Bubble Sort")
plt.plot(sizes, selection_times, marker="o", label="Selection Sort")
plt.xlabel("Jumlah Data")
plt.ylabel("Waktu Eksekusi (detik)")
plt.title("Bubble Sort vs Selection Sort")
plt.legend()
plt.grid(True)
plt.show()
````

Tampilkan mahasiswa dari nilai **terendah ke tertinggi** menggunakan Selection Sort.

````python
mahasiswa = [
    {"nama": "Andi", "nilai": 78},
    {"nama": "Budi", "nilai": 65},
    {"nama": "Citra", "nilai": 90},
    {"nama": "Dinda", "nilai": 55},
    {"nama": "Eka", "nilai": 88},
    {"nama": "Fajar", "nilai": 72},
]


def selection_sort_nilai(data):
    arr = data.copy()

    for i in range(len(arr) - 1):
        min_index = i

        for j in range(i + 1, len(arr)):
            if arr[j]["nilai"] < arr[min_index]["nilai"]:
                min_index = j

        arr[i], arr[min_index] = arr[min_index], arr[i]

    return arr


hasil = selection_sort_nilai(mahasiswa)

for item in hasil:
    print(item["nama"], item["nilai"])
````

## 13. Studi Kasus 3 — Data Hampir Terurut

Perhatikan:

```text
[10, 20, 30, 40, 60, 50]
```

Hanya ada satu pasangan yang salah.

Pertanyaan:

> Bubble Sort teroptimasi atau Selection Sort — mana yang lebih menarik untuk kondisi ini?

Mari hitung operasinya.

````python
data_hampir_urut = [10, 20, 30, 40, 60, 50]

def bubble_optimized_count(data):
    arr = data.copy()
    n = len(arr)
    comparisons = 0
    swaps = 0
    passes = 0

    for i in range(n - 1):
        swapped = False
        passes += 1

        for j in range(n - 1 - i):
            comparisons += 1

            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swaps += 1
                swapped = True

        if not swapped:
            break

    return arr, comparisons, swaps, passes


print("Bubble optimized:", bubble_optimized_count(data_hampir_urut))
print("Selection       :", selection_sort_count(data_hampir_urut))
````

### Insight

Bubble Sort teroptimasi bisa berhenti lebih cepat pada data yang sudah atau hampir terurut.

Selection Sort tidak memiliki keuntungan *early stop* yang sama karena tetap mencari minimum dari sisa data.

## 15. Stability

Sebuah sorting algorithm disebut **stable** jika data yang memiliki key sama tetap mempertahankan urutan relatif awalnya.

Contoh:

```text
(A, 80)
(B, 70)
(C, 80)
```

Jika sorting berdasarkan nilai, hasil stable seharusnya tetap menempatkan `A` sebelum `C`.

````python
data_nilai = [
    ("A", 80),
    ("B", 70),
    ("C", 80),
    ("D", 60),
]


def bubble_sort_tuple_by_score(data):
    arr = data.copy()

    for i in range(len(arr) - 1):
        for j in range(len(arr) - 1 - i):
            if arr[j][1] > arr[j + 1][1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

    return arr


print(bubble_sort_tuple_by_score(data_nilai))
````

Bubble Sort dapat bersifat **stable** jika pertukaran hanya dilakukan ketika:

```python
arr[j] > arr[j + 1]
```

dan bukan ketika nilainya sama.

Selection Sort versi sederhana umumnya **tidak stable** karena pertukaran dapat mengubah urutan relatif elemen yang memiliki key sama.

## 16. Kapan Tidak Menggunakan Keduanya?

Bubble Sort dan Selection Sort berguna untuk memahami dasar algoritma, tetapi untuk data besar biasanya bukan pilihan utama karena kompleksitasnya **O(n²)**.

Untuk data lebih besar, pada pertemuan selanjutnya kita akan melihat algoritma yang lebih efisien.
