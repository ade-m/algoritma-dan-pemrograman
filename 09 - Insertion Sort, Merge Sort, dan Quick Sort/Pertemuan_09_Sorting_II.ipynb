## 1. Fenomena: Semua sorting tidak sama

Misalkan kita memiliki data:

```text
[29, 10, 14, 37, 13]
```

Ketiga algoritma pada pertemuan ini dapat menghasilkan output yang sama:

```text
[10, 13, 14, 29, 37]
```

Tetapi **cara mencapai hasilnya berbeda**.

- Insertion Sort: membangun bagian kiri yang sudah terurut.
- Merge Sort: membagi data menjadi bagian kecil lalu menggabungkannya.
- Quick Sort: memilih pivot dan mempartisi data.

## 2. Insertion Sort — Analogi Menyusun Kartu

Bayangkan kita menyusun kartu di tangan.

Setiap kartu baru:
1. dibandingkan dengan kartu sebelumnya,
2. digeser sampai menemukan posisi yang tepat,
3. disisipkan ke posisi tersebut.

````python
def insertion_sort_trace(data):
    arr = data.copy()

    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        print(f"\nIterasi {i}")
        print("Sebelum :", arr)
        print("Key     :", key)

        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key
        print("Sesudah :", arr)

    return arr

data = [29, 10, 14, 37, 13]
hasil = insertion_sort_trace(data)
print("\nHasil akhir:", hasil)
````

### Kompleksitas Insertion Sort

| Kondisi | Kompleksitas |
|---|---|
| Best case | O(n) |
| Average case | O(n²) |
| Worst case | O(n²) |

**Best case** terjadi saat data hampir atau sudah terurut.

Insertion Sort juga:
- *in-place*,
- stabil,
- efektif untuk data kecil atau hampir terurut.

## 3. Merge Sort — Divide and Conquer

Merge Sort bekerja dengan dua tahap:

1. **Divide** — bagi data menjadi dua bagian terus-menerus.
2. **Merge** — gabungkan kembali dua bagian yang sudah terurut.

Contoh:

```text
[29, 10, 14, 37]

→ [29, 10]   [14, 37]
→ [29] [10]  [14] [37]

merge:
[10, 29]
[14, 37]

merge:
[10, 14, 29, 37]
```

````python
def merge(left, right):
    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])
    return result


def merge_sort(data):
    if len(data) <= 1:
        return data

    mid = len(data) // 2
    left = merge_sort(data[:mid])
    right = merge_sort(data[mid:])

    return merge(left, right)


data = [29, 10, 14, 37, 13]
print(merge_sort(data))
````

### Kompleksitas Merge Sort

| Kondisi | Kompleksitas |
|---|---|
| Best | O(n log n) |
| Average | O(n log n) |
| Worst | O(n log n) |

Trade-off utama:
- performa konsisten,
- tetapi membutuhkan ruang tambahan untuk proses merge.

## 4. Quick Sort — Pivot dan Partition

Quick Sort memilih satu elemen sebagai **pivot**, lalu membagi data menjadi:

```text
lebih kecil | pivot | lebih besar
```

Contoh:

```text
[29, 10, 14, 37, 13]
pivot = 29

kiri  = [10, 14, 13]
pivot = [29]
kanan = [37]
```

Proses yang sama diterapkan kembali pada bagian kiri dan kanan.

````python
def quick_sort(data):
    if len(data) <= 1:
        return data

    pivot = data[-1]

    left = [x for x in data[:-1] if x <= pivot]
    right = [x for x in data[:-1] if x > pivot]

    return quick_sort(left) + [pivot] + quick_sort(right)


data = [29, 10, 14, 37, 13]
print(quick_sort(data))
````

### Kompleksitas Quick Sort

| Kondisi | Kompleksitas |
|---|---|
| Best | O(n log n) |
| Average | O(n log n) |
| Worst | O(n²) |

Worst case dapat terjadi jika pemilihan pivot buruk secara berulang.

Quick Sort biasanya sangat cepat dalam praktik, tetapi performanya bergantung pada strategi partition dan pivot.

## 5. Bandingkan Ketiganya

| Algoritma | Best | Average | Worst | Stabil | Karakter utama |
|---|---:|---:|---:|---|---|
| Insertion Sort | O(n) | O(n²) | O(n²) | Ya | baik untuk data kecil/hampir terurut |
| Merge Sort | O(n log n) | O(n log n) | O(n log n) | Ya | performa konsisten |
| Quick Sort | O(n log n) | O(n log n) | O(n²) | Umumnya tidak | cepat dalam praktik |

## 6. Eksperimen Waktu Eksekusi

Bandingkan ketiga algoritma pada ukuran data yang berbeda.

````python
import random
import time

def insertion_sort(data):
    arr = data.copy()
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

sizes = [100, 500, 1000]

for n in sizes:
    data = [random.randint(1, 100000) for _ in range(n)]

    for name, func in [
        ("Insertion", insertion_sort),
        ("Merge", merge_sort),
        ("Quick", quick_sort)
    ]:
        start = time.perf_counter()
        func(data)
        elapsed = time.perf_counter() - start
        print(f"{name:10} | n={n:4} | {elapsed:.6f} detik")

    print("-" * 45)
````
