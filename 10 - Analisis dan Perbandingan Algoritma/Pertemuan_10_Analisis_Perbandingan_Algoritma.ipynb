## 1. Fenomena: Algoritma tercepat belum tentu selalu terbaik

Pertanyaan:

> Jika Binary Search lebih cepat daripada Linear Search, mengapa kita masih menggunakan Linear Search?

Jawabannya ada pada **kondisi data**.

Binary Search membutuhkan data terurut. Linear Search tidak.

Jadi evaluasi algoritma tidak cukup hanya melihat angka Big-O.

## 3. Searching: Linear vs Binary

```python
def linear_search(data, target):
    for i, value in enumerate(data):
        if value == target:
            return i
    return -1


def binary_search(data, target):
    low = 0
    high = len(data) - 1

    while low <= high:
        mid = (low + high) // 2

        if data[mid] == target:
            return mid
        elif data[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1
```

### Analisis Konseptual

| Faktor | Linear Search | Binary Search |
|---|---|---|
| Data harus terurut? | Tidak | Ya |
| Worst case | O(n) | O(log n) |
| Implementasi | sederhana | lebih kompleks |
| Data kecil | cukup baik | belum tentu perlu |
| Data besar terurut | kurang efisien | sangat baik |

## 4. Sorting: O(n²) vs O(n log n)

Bandingkan:
- Bubble Sort
- Selection Sort
- Insertion Sort
- Merge Sort
- Quick Sort

Pertanyaan penting bukan:

> "Algoritma mana paling cepat?"

Tetapi:

> "Algoritma mana paling sesuai untuk kasus ini?"

```python
import random
import time

def bubble_sort(data):
    arr = data.copy()
    n = len(arr)

    for i in range(n):
        swapped = False

        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True

        if not swapped:
            break

    return arr


def insertion_sort(data):
    arr = data.copy()

    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

    return arr
```
