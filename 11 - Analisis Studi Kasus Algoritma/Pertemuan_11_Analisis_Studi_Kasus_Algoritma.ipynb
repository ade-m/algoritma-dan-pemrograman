## 3. Studi Kasus 1 — Cari Pasien Berdasarkan Nomor Antrean

Data antrean:

```text
[101, 102, 103, 104, 105, 106]
```

User ingin mencari nomor antrean `104`.

Karena data terurut, kita dapat mempertimbangkan Binary Search.

````python
data_antrean = [101, 102, 103, 104, 105, 106]

def binary_search(data, target):
    low = 0
    high = len(data) - 1

    while low <= high:
        mid = (low + high) // 2

        if data[mid] == target:
            return mid
        elif data[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1

print(binary_search(data_antrean, 104))
````

### Diskusi

Apakah:
- diurutkan dahulu,
- gunakan Linear Search,
- atau ubah struktur data?

## 5. Studi Kasus 3 — Cari Harga Terdekat dengan Budget

Budget user:

```text
Rp 27.000
```

Daftar harga:

```text
10.000, 15.000, 20.000, 25.000, 30.000, 40.000
```

Target:

> harga terbesar yang masih <= budget.

````python
def harga_sesuai_budget(harga, budget):
    low = 0
    high = len(harga) - 1
    best = None

    while low <= high:
        mid = (low + high) // 2

        if harga[mid] <= budget:
            best = harga[mid]
            low = mid + 1
        else:
            high = mid - 1

    return best

harga = [10000, 15000, 20000, 25000, 30000, 40000]
print(harga_sesuai_budget(harga, 27000))
````

## 6. Studi Kasus 4 — Deteksi Duplikasi

Diberikan:

```python
data = [10, 4, 7, 10, 2, 4]
```

Tujuan:
- temukan nilai duplikat.

Diskusikan alternatif:
- nested loop,
- sorting,
- set/dictionary.

````python
data = [10, 4, 7, 10, 2, 4]

seen = set()
duplicates = set()

for x in data:
    if x in seen:
        duplicates.add(x)
    else:
        seen.add(x)

print("Duplikat:", duplicates)
````
