## 2. Contoh Sederhana — Nilai Kelulusan

Masalah:

> Tentukan apakah mahasiswa lulus jika nilai >= 60.

### IPO

**Input**
- nilai

**Process**
- bandingkan nilai dengan 60

**Output**
- "Lulus" atau "Tidak Lulus"

### Pseudocode

```text
START

INPUT nilai

IF nilai >= 60
    PRINT "Lulus"
ELSE
    PRINT "Tidak Lulus"

END
```

````python
def cek_kelulusan(nilai):
    if nilai >= 60:
        return "Lulus"
    return "Tidak Lulus"

print(cek_kelulusan(75))
print(cek_kelulusan(50))
````

## 3. Studi Kasus — Pencarian Produk

Masalah:

> User memasukkan kode produk. Sistem mencari apakah kode tersebut tersedia.

Pertanyaan:
1. Apakah data terurut?
2. Seberapa besar jumlah data?
3. Apakah pencarian sering dilakukan?
4. Linear Search atau Binary Search?

### Pseudocode Binary Search

```text
SET low = 0
SET high = n - 1

WHILE low <= high
    mid = (low + high) // 2

    IF data[mid] == target
        RETURN mid

    ELSE IF data[mid] < target
        low = mid + 1

    ELSE
        high = mid - 1

RETURN -1
```

````python
def binary_search(data, target):
    low = 0
    high = len(data) - 1

    while low <= high:
        mid = (low + high) // 2

        if data[mid] == target:
            return mid
        elif data[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1
````

## 4. Decomposition

Kasus:

> Sistem menerima daftar nilai mahasiswa dan harus menampilkan:
> - nilai tertinggi,
> - nilai terendah,
> - rata-rata,
> - jumlah mahasiswa lulus.

Pecah menjadi fungsi:
1. cari maksimum,
2. cari minimum,
3. hitung rata-rata,
4. hitung kelulusan.

````python
def ringkasan_nilai(data):
    tertinggi = max(data)
    terendah = min(data)
    rata_rata = sum(data) / len(data)
    jumlah_lulus = sum(1 for x in data if x >= 60)

    return {
        "tertinggi": tertinggi,
        "terendah": terendah,
        "rata_rata": rata_rata,
        "jumlah_lulus": jumlah_lulus
    }

nilai = [78, 90, 65, 88, 72, 55]
print(ringkasan_nilai(nilai))
````

## 6. Studi Kasus — Parkir

Masalah:

> Sistem menerima jam masuk dan jam keluar. Tarif:
> - 1 jam pertama: Rp5.000
> - setiap jam berikutnya: Rp3.000

Tugas:
1. buat IPO,
2. tulis pseudocode,
3. implementasikan.

````python
def biaya_parkir(jam):
    if jam <= 0:
        return 0

    if jam == 1:
        return 5000

    return 5000 + (jam - 1) * 3000

for jam in [1, 2, 5]:
    print(jam, "jam =", biaya_parkir(jam))
````
