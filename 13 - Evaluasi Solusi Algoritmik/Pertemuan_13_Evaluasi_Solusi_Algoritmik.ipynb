## 3. Contoh — Mencari Nilai Maksimum

Solusi A:

````python
def max_v1(data):
    terbesar = data[0]

    for x in data:
        if x > terbesar:
            terbesar = x

    return terbesar
````

Solusi B:

````python
def max_v2(data):
    return sorted(data)[-1]
````

Keduanya dapat menghasilkan jawaban sama.

Tetapi secara kompleksitas:

- `max_v1` → O(n)
- `max_v2` → O(n log n)

Jadi sorting seluruh data hanya untuk mencari maksimum adalah pekerjaan tambahan yang tidak perlu.

## 4. Benchmark Sederhana

````python
import random
import time

data = [random.randint(1, 1_000_000) for _ in range(200_000)]

start = time.perf_counter()
hasil1 = max_v1(data)
t1 = time.perf_counter() - start

start = time.perf_counter()
hasil2 = max_v2(data)
t2 = time.perf_counter() - start

print("Hasil sama?", hasil1 == hasil2)
print("max_v1:", t1)
print("max_v2:", t2)
````

## 5. Edge Cases

Pertanyaan:

Apa yang terjadi jika:

```python
data = []
```

Fungsi `max_v1()` akan gagal karena mengakses `data[0]`.

Solusi yang lebih robust perlu menangani input kosong.

````python
def max_safe(data):
    if not data:
        return None

    terbesar = data[0]

    for x in data[1:]:
        if x > terbesar:
            terbesar = x

    return terbesar

print(max_safe([]))
print(max_safe([3, 10, 7]))
````

## 9. Studi Kasus Evaluasi

Dua mahasiswa membuat solusi untuk mencari duplikasi.

### Solusi A
Nested loop.

### Solusi B
Menggunakan set.

Analisis:
- correctness,
- time complexity,
- memory,
- readability,
- konteks data kecil vs besar.

````python
def duplicate_nested(data):
    duplicates = set()

    for i in range(len(data)):
        for j in range(i + 1, len(data)):
            if data[i] == data[j]:
                duplicates.add(data[i])

    return duplicates


def duplicate_set(data):
    seen = set()
    duplicates = set()

    for x in data:
        if x in seen:
            duplicates.add(x)
        else:
            seen.add(x)

    return duplicates
````
